```python
import requests

URL = "https://www.scrapethissite.com/pages/simple/"

try: 
    headers= {
        "User-Agent" : "Mozilla/5.0"
    }
    
    res = requests.get(URL , headers = headers)
    res.raise_for_status()  #  Raise error for 4xx amd 5xx responses
    

except requests.exceptions.RequestException as e:
    print("Error: ", e)
```

```python


with open("scraped_data/data1.html", "w" , encoding= "utf-8") as f:
    f.write(res.text)
    print("Successfully Saved")
```

```python
#Beautifulsoup
from bs4 import BeautifulSoup 

with open("WEB SCRAPING/scraped_data/data1.html" , "r") as f:
    html_content = f.read()

soup= BeautifulSoup(html_content , "lxml")
```

```python
all_h3 = soup.find_all("h3")

all_countries = []
for h3 in all_h3 :
    name= h3.get_text(strip="True")
    # print(h3.find_parent("div")["class"])
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    area = h3.find_next("div").select_one("span.country-area").get_text(strip=True)
    all_countries.append([name , population , area])
```

```python
import pandas as pd

df = pd.DataFrame(all_countries , columns = ["Country" , "Population" , "Area"])
```

```python


df.to_csv("WEB SCRAPING/cleaned_data/data2.csv" , index = False)
```
