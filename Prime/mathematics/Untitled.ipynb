```python
from scipy.stats import binom
n = 5
k = 3
p = 0.5

prob = binom.pmf(k , n , p)
print(prob)
```

```python
#Uniform Distributions

import numpy as np 
import matplotlib.pyplot as plt

values = np.random.uniform(0, 10 , 10_00_000)

plt.hist(values , bins=100 , density = True , alpha = 0.3)
plt.title("Uniform Distributions")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid(True)

plt.show()
```

```python
#Normal Distribution

import numpy as np 
import matplotlib.pyplot as plt
from scipy.stats import norm

mu = 70
sigma = 10

#x values

x = np.linspace(30, 110 , 1000)

#Normal PDF 

y = norm.pdf(x , mu , sigma)

#plotting

plt.plot(x , y , color = 'black' , linewidth = 2 , label = "Normal Distribution")

#shadding 68% of region

x1 = np.linspace(mu-sigma , mu+sigma , 1000)
plt.fill_between(x1 , norm.pdf(x1 , mu , sigma) , color = 'green' ,alpha = 0.3 ,  label= "68% region")

#shading 95% region 

x2 = np.linspace(mu-2*sigma , mu+2*sigma , 1000)
plt.fill_between(x2 , norm.pdf(x2 , mu , sigma) , color = 'red' ,alpha = 0.2 ,  label= "95% region")

#shading 99.7% region 

x3 = np.linspace(mu-3*sigma , mu+3*sigma , 1000)
plt.fill_between(x3 , norm.pdf(x3 , mu , sigma) , color = 'blue' ,alpha = 0.1 ,  label= "99.7% region")


plt.xlabel("X")
plt.ylabel("Probability Distribution Values")
plt.legend()
plt.show()
```
