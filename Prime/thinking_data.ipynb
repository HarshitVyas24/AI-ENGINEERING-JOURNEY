```python
import json
```

```python
def load_data(filename):
    with open(filename, "r" ) as f:
        data = json.load(f)

    return data
    
```

```python
data = load_data("store_data.json")
print(data

```

```python
#clean and structure datt

def clean_data(data):
    cleaned_data = []
    unique_user = set()
    for user in data :


        text_to_num = {"one" : 1, "two" : 2, "three" : 3, "four" : 4, "five" : 5, "5" : 5, "3.5" : 3.5,}
         #clean Ratings
        raw_ratings = str(user["rating"]).strip().lower()
       
        if (raw_ratings in text_to_num):
            raw_ratings = text_to_num[raw_ratings]
        
        user["rating"] = int(raw_ratings)
        #Handling the missing value
        
        if ("age" not in user):
            user["age"] = None

        if(user["name"] in unique_user):
           continue
        unique_user.add(user["name"])
            
        cleaned_data.append(user)

    return cleaned_data
           
           
        
```

```python
# clean_data(data)
data = clean_data(data)
print(data)
```

```python
#Avg Rating Calculation
def get_insights(data):
    sum = 0
    count = 0
    poor_rating = []
    for user in data:
        if (user["rating"]<3):
            poor_rating.append(user)
        sum+= user["rating"]
        count+=1
        
    
    avg_rating = sum/count
    percentage_poor_rating = (len(poor_rating)*100)/count
    
    print(f"On an average the users have rated the Store a rating of {avg_rating} ")
    print(f"{percentage_poor_rating}% of users rate the store poorly.")
```

```python
get_insights(data)
```

```python
#Build recommendation Feature

def get_recommendation(data):
    recommended_brand = {}
    for user in data:
        if (user["rating"]>4):
            recommended_brand[user["name"]] = "Apple"
        else:
            recommended_brand[user["name"]] = "Samsung"

    return recommended_brand
```

```python
get_recommendation(data)
```
